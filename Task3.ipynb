```python
import pandas as pd

# 1. Load CSV (fix Unicode error)
df = pd.read_csv("C:\Users\AASMI NISHA M\Downloads\sales_dataset.csv"

# 2. Convert Quantity & Price to numbers (fix TypeError)
df["Quantity"] = pd.to_numeric(df["Quantity"], errors="coerce")
df["Price"] = pd.to_numeric(df["Price"], errors="coerce")

# 3. Replace missing values
df = df.fillna(0)

# 4. Calculate Total
df["Total"] = df["Quantity"] * df["Price"]

# 5. Total Sales
total_sales = df["Total"].sum()

# 6. Best-selling product by quantity
best_product = df.loc[df["Quantity"].idxmax(), "Product"]

# 7. Print Report
print("===== SALES REPORT =====")
print("Total Sales:", total_sales)
print("Best-Selling Product:", best_product)
print("\nDataset Preview:")
print(df.head())
```
